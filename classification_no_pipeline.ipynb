# Classification Track — Insurance Claims Fraud Detection (no-`Pipeline` variant)
**23CSE301 Machine Learning Capstone — Review 1 (Part A)**

**Dataset:** `data/insurance_claims/insurance_claims.csv` (1,000 rows x 40 columns) — auto-insurance policy and incident records.

**Problem statement:** Predict **`fraud_reported`** (Yes/No) — whether an insurance claim is fraudulent — from policy, customer and incident attributes.

**How this notebook differs from `classification.ipynb`:** the original notebook wraps preprocessing + model into a single `sklearn.pipeline.Pipeline` object per algorithm and loops over a `{name: pipeline}` dict to train all five. Here, preprocessing (imputation, encoding, scaling, PCA) is done **once, manually**, step by step, producing plain NumPy arrays (`X_train_final`, `X_test_final`) — no `ColumnTransformer`, no `Pipeline`. Each of the 5 Part-A algorithms then gets **its own standalone cell** that directly calls `.fit()` / `.predict()` on those arrays. This makes each algorithm's code independently readable and editable, at the cost of repeating the fit/evaluate/print boilerplate five times instead of once in a loop.

**Scope of this notebook (Review 1, Part A):**

1. Data loading & audit
2. Exploratory Data Analysis (EDA)
3. Data cleaning, feature engineering
4. Manual encoding, scaling & train/test split (no `ColumnTransformer`)
5. **Dimensionality reduction (PCA)** — fit on the training split, applied manually (no `Pipeline`) to produce the final feature matrices used by every Part-A algorithm
6. **One dedicated cell per Part-A algorithm** — Logistic Regression, K-Nearest Neighbors, Naive Bayes (Gaussian), Decision Tree Classifier, Support Vector Machine (SVC) — each fit directly on `X_train_final`/`y_train`
7. Preliminary comparison table (Accuracy, weighted F1, confusion matrix, ROC-AUC)

**Part B** (Random Forest, AdaBoost, Gradient Boosting, Bagging, MLP) and the final 10-algorithm consolidated comparison are out of scope for Review 1.

**Random seed** `42` is used everywhere for reproducibility.


```python
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.decomposition import PCA
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.svm import SVC
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                              confusion_matrix, ConfusionMatrixDisplay, roc_auc_score, roc_curve)

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

sns.set_theme(style='whitegrid', palette='colorblind')
plt.rcParams['figure.dpi'] = 100

DATA_PATH = '../data/insurance_claims/insurance_claims.csv'
```

## Section B — Preprocessing & Feature Engineering

### B1 — Data cleaning

```python
df_clean = df.copy()

# Drop the empty artifact column and pure-identifier columns that carry no generalisable signal.
df_clean = df_clean.drop(columns=['_c39', 'policy_number', 'insured_zip', 'incident_location'])

# Outlier check (IQR rule) on key monetary numeric columns.
outlier_check_cols = ['total_claim_amount', 'policy_annual_premium', 'injury_claim', 'property_claim', 'vehicle_claim']
for col in outlier_check_cols:
    q1, q3 = df_clean[col].quantile([0.25, 0.75])
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    n_out = ((df_clean[col] < lower) | (df_clean[col] > upper)).sum()
    print(f'{col}: {n_out} outliers by IQR rule ({n_out / len(df_clean):.2%})')

print('\nMissing values (collision_type, property_damage, police_report_available, authorities_contacted)')
print('will be handled with a most-frequent-category imputer fitted on the training split only (see B2')
print('pipeline) — dropping those rows would lose too much data for a 1,000-row dataset.')
print('\nOutlier counts above are low (<3% each); genuine large claims are a real and informative')
print('part of a fraud-detection problem, so they are kept as-is (not capped or removed).')
```

### B3 — Feature engineering

```python
# Feature 1: claim_to_premium_ratio — how large the claim is relative to what the customer
# actually pays in annual premium. A claim that dwarfs the policyholder's premium is a classic
# red flag insurers use manually, and it is not present as a single ready-made column.
df_clean['claim_to_premium_ratio'] = df_clean['total_claim_amount'] / (df_clean['policy_annual_premium'] + 1)

# Feature 2: days_bind_to_incident — number of days between the policy being bound and the
# incident occurring. Claims filed very soon after a policy starts ("early claims") are a
# well-known fraud indicator in the insurance industry, so this is engineered explicitly from
# the two raw date columns before those dates are dropped as high-cardinality raw strings.
df_clean['policy_bind_date'] = pd.to_datetime(df_clean['policy_bind_date'])
df_clean['incident_date'] = pd.to_datetime(df_clean['incident_date'])
df_clean['days_bind_to_incident'] = (df_clean['incident_date'] - df_clean['policy_bind_date']).dt.days

df_clean = df_clean.drop(columns=['policy_bind_date', 'incident_date'])

print(df_clean[['claim_to_premium_ratio', 'days_bind_to_incident']].describe())
```

### B2 — Encoding, scaling & train/test split (manual, no `ColumnTransformer`)

- **Target encoding:** `fraud_reported` → 1 ("Y", fraud) / 0 ("N", not fraud).
- **Categorical columns** → one-hot encoded; the 4 columns with missing values are first imputed with the most-frequent category.
- **Numeric columns** → standard-scaled.
- A **stratified** 80:20 train/test split is used (`stratify=y`) first, then every imputer/encoder/scaler below is `.fit()` on the training split only and `.transform()`-only on the test split — no leakage — but each step is called explicitly, one at a time, instead of being wrapped in a `ColumnTransformer`/`Pipeline`.

```python
y = (df_clean['fraud_reported'] == 'Y').astype(int)
X = df_clean.drop(columns=['fraud_reported'])

cat_features = X.select_dtypes(include='object').columns.tolist()
num_features = [c for c in X.columns if c not in cat_features]

cat_missing = ['collision_type', 'property_damage', 'police_report_available', 'authorities_contacted']
cat_complete = [c for c in cat_features if c not in cat_missing]

# 1. Split FIRST, so every fit below only ever sees the training rows.
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y)

# 2. Impute the 4 categorical columns that have missing values (fit on train only).
cat_imputer = SimpleImputer(strategy='most_frequent')
X_train_cat_missing = pd.DataFrame(cat_imputer.fit_transform(X_train[cat_missing]),
                                    columns=cat_missing, index=X_train.index)
X_test_cat_missing = pd.DataFrame(cat_imputer.transform(X_test[cat_missing]),
                                   columns=cat_missing, index=X_test.index)

# 3. Recombine all 17 categorical columns and one-hot encode them together (fit on train only).
X_train_cat_all = pd.concat([X_train[cat_complete], X_train_cat_missing], axis=1)[cat_features]
X_test_cat_all = pd.concat([X_test[cat_complete], X_test_cat_missing], axis=1)[cat_features]

encoder = OneHotEncoder(handle_unknown='ignore', drop='first', sparse_output=False)
X_train_cat_encoded = encoder.fit_transform(X_train_cat_all)
X_test_cat_encoded = encoder.transform(X_test_cat_all)

# 4. Standard-scale the numeric columns (fit on train only).
scaler = StandardScaler()
X_train_num_scaled = scaler.fit_transform(X_train[num_features])
X_test_num_scaled = scaler.transform(X_test[num_features])

# 5. Concatenate numeric + categorical blocks into the final encoded matrices.
X_train_enc = np.hstack([X_train_num_scaled, X_train_cat_encoded])
X_test_enc = np.hstack([X_test_num_scaled, X_test_cat_encoded])

print(f'Train: {X_train_enc.shape}, Test: {X_test_enc.shape}')
print(f'{len(num_features)} numeric features, {len(cat_features)} categorical features (before one-hot expansion)')
print(f'Train class balance: {y_train.mean():.3f}   Test class balance: {y_test.mean():.3f}')
```

## Section C — Dimensionality Reduction (PCA)

One-hot encoding 17 categorical columns (several high-cardinality, e.g. `insured_hobbies`, `insured_occupation`) considerably expands the raw 34-column feature set. Before training any Part-A algorithm, PCA is fit **on the training split only** (no leakage — the fitted transform is then applied to the test split), and the resulting reduced representation becomes the shared input for every algorithm in Section D below.

### C1 — Fit PCA on the encoded training features & choose the number of components

```python
print(f'Encoded feature matrix: {X_train_enc.shape[1]} features (from {X_train.shape[1]} raw columns, after one-hot expansion)')

pca_full = PCA(random_state=RANDOM_STATE)
pca_full.fit(X_train_enc)

cum_var = np.cumsum(pca_full.explained_variance_ratio_)
n90 = int(np.argmax(cum_var >= 0.90) + 1)
n95 = int(np.argmax(cum_var >= 0.95) + 1)
print(f'Components needed for 90% variance: {n90}')
print(f'Components needed for 95% variance: {n95}')

plt.figure(figsize=(9, 5.5))
plt.plot(range(1, len(cum_var) + 1), cum_var, marker='.')
plt.axhline(0.90, color='green', linestyle='--', label='90% variance')
plt.axhline(0.95, color='red', linestyle='--', label='95% variance')
plt.axvline(n90, color='green', linestyle=':')
plt.axvline(n95, color='red', linestyle=':')
plt.xlabel('Number of principal components')
plt.ylabel('Cumulative explained variance')
plt.title('PCA Scree Curve — Insurance Claims Features')
plt.legend()
plt.tight_layout()
plt.savefig('../models/clf_nopipe_pca_scree.png', bbox_inches='tight')
plt.show()
```

**Observation:** One-hot encoding expands the raw 35-column feature set to **142 encoded features**. Variance decays slowly — **55 of 142 components (≈39%)** are needed for 90% variance and **75 (≈53%)** for 95%, a direct consequence of one-hot-encoding many rare categories into near-orthogonal sparse dimensions. `n90 = 55` is used below as the shared dimensionality for every Part-A algorithm.

### C2 — 2D PCA projection, coloured by `fraud_reported`

```python
pca_2d = PCA(n_components=2, random_state=RANDOM_STATE)
X_train_pca2 = pca_2d.fit_transform(X_train_enc)

plt.figure(figsize=(8, 6.5))
for cls, label, color in [(0, 'N (not fraud)', '#4C72B0'), (1, 'Y (fraud)', '#C44E52')]:
    mask = y_train.values == cls
    plt.scatter(X_train_pca2[mask, 0], X_train_pca2[mask, 1], s=18, alpha=0.6, label=label, color=color)
plt.xlabel(f'PC1 ({pca_2d.explained_variance_ratio_[0]:.1%} var)')
plt.ylabel(f'PC2 ({pca_2d.explained_variance_ratio_[1]:.1%} var)')
plt.title('2D PCA Projection — coloured by fraud_reported')
plt.legend()
plt.tight_layout()
plt.savefig('../models/clf_nopipe_pca_2d_scatter.png', bbox_inches='tight')
plt.show()

print(f'PC1 + PC2 together explain only {pca_2d.explained_variance_ratio_.sum():.1%} of total variance.')
```

**Observation:** PC1+PC2 together capture only 23.7% of total variance, and fraud/non-fraud points overlap heavily with no visually separable cluster — consistent with the EDA finding that no small set of features cleanly separates the two classes. `n90 = 55` components (from C1) are used from here on, not just these 2, as the actual modelling dimensionality.

### C3 — Apply PCA to get the final training/test feature matrices

```python
# No Pipeline wrapper — PCA is fit directly on the already-encoded/scaled training matrix
# (fit on train only, per the no-leakage rule) and every Part-A algorithm below is fit
# directly on the resulting X_train_final / X_test_final NumPy arrays.
pca = PCA(n_components=n90, random_state=RANDOM_STATE)
X_train_final = pca.fit_transform(X_train_enc)
X_test_final = pca.transform(X_test_enc)

print(f'Final modelling matrices: {X_train.shape[1]} raw columns -> {X_train_enc.shape[1]} encoded -> {n90} PCA components.')
print(f'X_train_final: {X_train_final.shape}, X_test_final: {X_test_final.shape}')
```

## Section D — Classification Track, Part A: Implementation (D1)

Five Part-A algorithms are trained: **Logistic Regression, K-Nearest Neighbors, Naive Bayes (Gaussian), Decision Tree Classifier, Support Vector Machine (SVC)**. Unlike `classification.ipynb`, there is **no `Pipeline` and no dict-loop** here — each algorithm gets its own standalone cell that calls `.fit(X_train_final, y_train)` directly. All five still train on the exact same 55-dimensional PCA-reduced representation built in Section C, so the comparison remains fair.

Results accumulate into shared `results_a` / `predictions_a` / `fitted_models_a` containers, initialised once below, so the downstream confusion-matrix/ROC/comparison-table cells work unchanged regardless of which algorithm's cell was run.

```python
# Shared result containers — re-run this cell to reset before re-running the 5 algorithm
# cells below in any order.
results_a = []
predictions_a = {}
fitted_models_a = {}

def evaluate_and_record(name, model, X_tr, y_tr, X_te, y_te):
    """Fit `model` on the training data, evaluate on the test data, and record results."""
    model.fit(X_tr, y_tr)
    preds = model.predict(X_te)
    proba = model.predict_proba(X_te)[:, 1]

    acc = accuracy_score(y_te, preds)
    prec = precision_score(y_te, preds, average='weighted', zero_division=0)
    rec = recall_score(y_te, preds, average='weighted', zero_division=0)
    f1 = f1_score(y_te, preds, average='weighted', zero_division=0)
    auc = roc_auc_score(y_te, proba)

    results_a.append({'Model': name, 'Accuracy': acc, 'Precision': prec, 'Recall': rec, 'F1 (weighted)': f1, 'ROC-AUC': auc})
    predictions_a[name] = (preds, proba)
    fitted_models_a[name] = model
    print(f'{name:<30s}  Acc={acc:.4f}  Prec={prec:.4f}  Rec={rec:.4f}  F1={f1:.4f}  AUC={auc:.4f}')
    return model

print('Result containers initialised.')
```

### D1.1 — Logistic Regression

```python
log_reg = LogisticRegression(max_iter=2000, random_state=RANDOM_STATE)
evaluate_and_record('Logistic Regression', log_reg, X_train_final, y_train, X_test_final, y_test)
```

### D1.2 — K-Nearest Neighbors

```python
knn = KNeighborsClassifier(n_neighbors=9)
evaluate_and_record('K-Nearest Neighbors', knn, X_train_final, y_train, X_test_final, y_test)
```

### D1.3 — Naive Bayes (Gaussian)

```python
nb = GaussianNB()
evaluate_and_record('Naive Bayes (Gaussian)', nb, X_train_final, y_train, X_test_final, y_test)
```

### D1.4 — Decision Tree Classifier

```python
dec_tree = DecisionTreeClassifier(max_depth=6, random_state=RANDOM_STATE)
evaluate_and_record('Decision Tree Classifier', dec_tree, X_train_final, y_train, X_test_final, y_test)
```

### D1.5 — Support Vector Machine (SVC)

```python
svc = SVC(C=1.0, kernel='rbf', probability=True, random_state=RANDOM_STATE)
evaluate_and_record('Support Vector Machine (SVC)', svc, X_train_final, y_train, X_test_final, y_test)

print('\nAll 5 Part-A classification algorithms trained successfully with no errors.')
```
